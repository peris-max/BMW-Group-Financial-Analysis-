```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    r"C:\Users\perik\OneDrive\Έγγραφα\bmw_financials.csv"
)
```

```python
print(df.head())
print(df.info())
print(df.columns)
```

```python
# Revenue growth
df["Revenue_Growth"] = (
    df["Revenue"].pct_change() * 100
).round(2)

# EBIT margin
df["EBIT_Margin"] = (
    df["EBIT"] / df["Revenue"] * 100
).round(2)

# Net profit margin
df["Net_Margin"] = (
    df["Net_Profit"] / df["Revenue"] * 100
).round(2)

# Return on Assets
df["ROA"] = (
    df["Net_Profit"] / df["Total_Assets"] * 100
).round(2)

# Return on Equity
df["ROE"] = (
    df["Net_Profit"] / df["Equity"] * 100
).round(2)

# Operating Cash Flow Margin
df["OCF_Margin"] = (
    df["Operating_Cash_Flow"] / df["Revenue"] * 100
).round(2)
```

```python
analysis = df[[
    "Year",
    "Revenue",
    "Revenue_Growth",
    "EBIT",
    "EBIT_Margin",
    "Net_Profit",
    "Net_Margin",
    "ROA",
    "ROE",
    "Operating_Cash_Flow",
    "OCF_Margin",
    "Cash"
]]
```

```python
print(analysis)
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["Year"],
    df["Revenue"],
    marker="o"
)

plt.title("BMW Group Revenue (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Revenue (€ million)")
plt.xticks(df["Year"])
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["Year"],
    df["EBIT_Margin"],
    marker="o",
    label="EBIT Margin"
)

plt.plot(
    df["Year"],
    df["Net_Margin"],
    marker="o",
    label="Net Margin"
)

plt.title("BMW Group Profitability (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Margin (%)")
plt.xticks(df["Year"])
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["Year"],
    df["ROA"],
    marker="o",
    label="ROA"
)

plt.plot(
    df["Year"],
    df["ROE"],
    marker="o",
    label="ROE"
)

plt.title("BMW Group ROA vs ROE (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Return (%)")
plt.xticks(df["Year"])
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["Year"],
    df["OCF_Margin"],
    marker="o"
)

plt.title("BMW Group Operating Cash Flow Margin (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Operating Cash Flow Margin (%)")
plt.xticks(df["Year"])
plt.grid(alpha=0.3)

plt.show()
```
